```python
import pandas as pd

# Load datasets
movies = pd.read_csv('../data/tmdb_5000_movies.csv')
tracks = pd.read_csv('../data/tracks.csv')

#Datasets that i am working with
print("Movies shape:", movies.shape)
print("Tracks shape:", tracks.shape)
movies.head()
```

```python
print("Movie columns:")
print(movies.columns.tolist())

print("\nTrack columns:")
print(tracks.columns.tolist())
```

```python
# Keeping only useful columns
movies = movies[['title', 'genres', 'vote_average', 'overview']]
tracks = tracks[['track_name', 'artists', 'track_genre', 'valence', 'energy', 'tempo', 'danceability']]

print("Movies cleaned:")
print(movies.head(3))
print("\nTracks cleaned:")
print(tracks.head(3))
```
