```python
import pandas as pd

# Load datasets
movies = pd.read_csv('../data/tmdb_5000_movies.csv')
tracks = pd.read_csv('../data/tracks.csv')

#Datasets that i am working with
print("Movies shape:", movies.shape)
print("Tracks shape:", tracks.shape)
movies.head()
```

```python
print("Movie columns:")
print(movies.columns.tolist())

print("\nTrack columns:")
print(tracks.columns.tolist())
```

```python
# Keeping only useful columns
movies = movies[['title', 'genres', 'vote_average', 'overview']]
tracks = tracks[['track_name', 'artists', 'track_genre', 'valence', 'energy', 'tempo', 'danceability']]

print("Movies cleaned:")
print(movies.head(3))
print("\nTracks cleaned:")
print(tracks.head(3))
```

```python
# Define mood profiles
mood_profiles = {
    'happy':     {'valence': 0.8, 'energy': 0.7, 'danceability': 0.7, 'tempo': 120},
    'sad':       {'valence': 0.2, 'energy': 0.3, 'danceability': 0.3, 'tempo': 70},
    'energetic': {'valence': 0.7, 'energy': 0.9, 'danceability': 0.8, 'tempo': 140},
    'chill':     {'valence': 0.5, 'energy': 0.3, 'danceability': 0.4, 'tempo': 90},
    'romantic':  {'valence': 0.6, 'energy': 0.4, 'danceability': 0.5, 'tempo': 100},
}

# Define which movie genres match each mood
mood_genres = {
    'happy':     ['Comedy', 'Animation', 'Family'],
    'sad':       ['Drama', 'Romance'],
    'energetic': ['Action', 'Thriller', 'Adventure'],
    'chill':     ['Documentary', 'Music', 'Fantasy'],
    'romantic':  ['Romance', 'Drama', 'Comedy'],
}

print("Mood profiles defined!")
print(list(mood_profiles.keys()))
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Normalize the track audio features
scaler = MinMaxScaler()
track_features = tracks[['valence', 'energy', 'danceability', 'tempo']].copy()
track_features_scaled = scaler.fit_transform(track_features)

print("Tracks scaled successfully!")
print("Shape:", track_features_scaled.shape)
```

```python
def recommend_songs(mood, n=5):
    # Get the mood profile
    mood_vector = pd.DataFrame([mood_profiles[mood]])
    mood_vector_scaled = scaler.transform(mood_vector)
    
    # Calculate cosine similarity between mood and all tracks
    similarities = cosine_similarity(mood_vector_scaled, track_features_scaled)
    
    # Get top n most similar tracks
    top_indices = similarities[0].argsort()[-n:][::-1]
    
    results = tracks.iloc[top_indices][['track_name', 'artists', 'track_genre']].copy()
    results['similarity'] = similarities[0][top_indices]
    
    return results.reset_index(drop=True)

# Test it!
print("Testing happy mood:")
print(recommend_songs('happy'))
```

```python
from langdetect import detect
from sklearn.preprocessing import MinMaxScaler

# Step 1: Fast genre filtering
english_genres = [
    'pop', 'rock', 'indie', 'jazz', 'blues', 'country',
    'hip-hop', 'r-n-b', 'soul', 'disco', 'edm',
    'house', 'electronic', 'folk', 'classical',
    'metal', 'punk', 'alternative', 'dance'
]

tracks_filtered = tracks[
    tracks['track_genre'].isin(english_genres)
].copy()

print("After genre filtering:", len(tracks_filtered))


# Step 2: Language detection
def is_english(text):
    try:
        return detect(str(text)) == 'en'
    except:
        return False


# Apply language detection ONLY on filtered tracks
tracks_english = tracks_filtered[
    tracks_filtered['track_name'].apply(is_english)
].copy()

print("Final English tracks:", len(tracks_english))


# Step 3: Feature extraction
track_features_english = tracks_english[
    ['valence', 'energy', 'danceability', 'tempo']
].copy()


# Step 4: Scaling
scaler = MinMaxScaler()

track_features_english_scaled = scaler.fit_transform(
    track_features_english
)

print("Scaling complete!")
tracks_english.to_csv('../data/tracks_english.csv', index=False)
print("Saved!")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_songs(mood, n=5):

    # Create mood vector
    mood_vector = pd.DataFrame([mood_profiles[mood]])

    # Scale mood vector using SAME scaler
    mood_vector_scaled = scaler.transform(mood_vector)

    # Compute cosine similarity
    similarities = cosine_similarity(
        mood_vector_scaled,
        track_features_english_scaled
    )

    # Get top 50 most similar songs
    top_indices = similarities[0].argsort()[-50:][::-1]

    # Fetch song details
    results = tracks_english.iloc[top_indices][
        ['track_name', 'artists', 'track_genre']
    ].copy()

    # Add similarity scores
    results['similarity'] = similarities[0][top_indices]

    # Remove duplicate tracks
    results = results.drop_duplicates(subset='track_name')

    # Remove repeated artists
    results = results.drop_duplicates(subset='artists')

    # Return top N recommendations
    return results.head(n).reset_index(drop=True)


# Test recommendation
print(recommend_songs('happy'))
```
